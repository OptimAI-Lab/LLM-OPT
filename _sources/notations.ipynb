# Introduction to Optimization



In this section we provide notations used throughput this book. 


\begin{itemize}
\item Real: $\RR$; complex: $\CC$; integer: $\ZZ$.
\item Scalars: $a,b,d\ldots$ ($c$ is reserved for constants)
\item Vectors: $\va,\vb,\ldots$ and they are \emph{column} vectors. Vector $\vn$ is reserved for noise.
\item $\v0 = [0,\cdots, 0]^T$ and $\v1 = [1,\cdots, 1]^T$
\item  $\sign(x)\in\{-1,0,1\}$ is the sign of $x$. $\sign(\vx) = [\sign(x_1),\ldots,\sign(x_N)]^T$
\item The transpose of a matrix $\vA$ is $\vA^T$. The conjugate transpose is $A^*$.
\item The set of $N$-by-$N$ symmetric matrices: $\SS^N$; symmetric positive semi-definite: $\SSp^N$; symmetric positive definite: $\SSpp^N$
\end{itemize}
