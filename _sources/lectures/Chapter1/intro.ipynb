# Introduction to Optimization
Optimization is the science of making the best possible decision from a set of available options. It arises naturally in everyday life:

- A company wants to maximize profit while minimizing costs.
- A commuter wants to choose the fastest route under traffic conditions.
- A researcher wants to allocate limited computational resources to various tasks.

At its core, optimization is about trade-offs: between objectives, between feasibility and optimality, and between computational cost and solution quality.

Mathematically, an optimization problem seeks to minimize (or maximize) an objective function over a set of variables, often under constraints:

```{math}
\begin{aligned}
\min_{x \in \mathcal{X}} \quad & f(x)
\end{aligned}
```
$x$: decision variable (can be a scalar, vector, or even structured object)

$f(x)$: objective or cost function, often measuring prediction error, energy, or resource usage

$\mathcal{X}$: feasible set, describing constraints on $x$

This general framework encompasses a vast array of real-world problems in machine learning, signal processing, control, and beyond.

## Why Optimization?
Optimization appears everywhere in machine learning and artificial intelligence:

Training a linear regression model: minimizing squared error

Training neural networks: minimizing a cross-entropy loss

Aligning LLMs: optimizing reward functions under preference constraints

Adversarial robustness: maximizing model loss with respect to small input perturbations

These problems differ in their structure, but they all boil down to choosing parameters to minimize a loss (or maximize a reward), possibly under various constraints.

## Example 1: Linear Regression
One of the most basic and widely used optimization problems is least squares regression:

```{math}
\min_{\vw \in \RR^d} \quad \frac{1}{2n} \sum_{i=1}^n (\vw^\top \va_i - b_i)^2
```
Here, each data point $(\va_i, b_i)$ contains features and a scalar label. The goal is to find a weight vector $\vx$ that best predicts $b_i$ from $\va_i$. This problem admits a closed-form solution but also provides a clean example of differentiable, unconstrained optimization.

## Example 2: Support Vector Machine (SVM)
Consider a binary classification problem with labels $b_i \in {-1, 1}$. The hard-margin SVM seeks the maximum-margin separating hyperplane:

```{math}
\begin{aligned}
\min_{\vw \in \RR^d} \quad & \|\vw\|^2 \\
\text{s.t.} \quad & b_i (\vx^\top \va_i + x_0) \ge 1, \quad \forall i
\end{aligned}
```
This is a convex, constrained optimization problem. It introduces duality theory and the idea of optimization under structural constraints — core ideas relevant to many modern methods including fine-tuning and prompt optimization for LLMs.

## Example 3: Training Neural Networks
Mathematically, an optimization problem is written as:

```{math}
\begin{aligned}
\min_{\vw \in \mathcal{W}} \quad f(\vw)
\end{aligned}
```
$\vw$ is the optimization variable (e.g., model weights or parameters).

$f(\vw)$ is the objective function — a scalar function we wish to minimize (e.g., loss, cost, error).

$\mathcal{W}$ is the feasible set — a set of constraints on $\vw$, such as non-negativity, norm bounds, or sparsity.

Key questions include:

- Does a solution $\vw^*$ exist?
- If so, how do we characterize and compute it?
- Can we guarantee its quality, efficiency, or robustness?

Optimization theory offers a unified language and set of tools to tackle these questions across a wide range of domains.

## Example 4: Adversarial Attacks
In adversarial learning, we formulate an optimization problem to maximize model error via perturbations:

```{math}
\begin{aligned}
\max_{\vdelta} \quad & \ell(\vx + \vdelta; y) \\
\text{s.t.} \quad & \|\vdelta\| \le \epsilon
\end{aligned}
```
This highlights a robust optimization setting, where we aim to find worst-case inputs within a constrained region. The dual of this idea appears in robust LLM alignment and safety tuning.

## Elements of an Optimization Problem
Every optimization problem is characterized by:

- The type of variables: continuous, discrete, mixed
- The objective function: smooth/nonsmooth, convex/nonconvex, deterministic/stochastic
- The constraints: equalities, inequalities, box, norm, combinatorial
- The scale: number of variables and data points (relevant to LLMs!)
- The desired accuracy and computation budget

Understanding these factors helps guide both the modeling and the algorithm selection process.
