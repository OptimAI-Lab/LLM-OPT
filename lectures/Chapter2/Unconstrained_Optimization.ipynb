# Section 2.1: Unconstrained Optimization



## 2.1.1 Unconstrained Optimization: Definitions and First Principles
In this chapter, we begin our study of optimization algorithms by focusing on the unconstrained case, where the optimization variable can be chosen from the entire space  $\RR^d$. That is, $\vw$ can be chosen from the entire $\RR^d$ space.


### Problem Setup
The general form of an unconstrained optimization problem is:

```{math}
\min_{\vw \in \RR^d} \quad f(\vw)
```
where $\vw$ is the optimization variable; $f: \RR^d \to \RR$ is the objective function; There are no explicit constraints on $\vw$. 

### Local and Global Optimal Solutions

````{admonition} Definition: Local and Global Optimal Solutions
:class: tip

Let $f: \RR^d \to \RR$ be a continuous function.

- **Local Minimum**: $\vw^*$ is a local minimizer if there exists $\epsilon > 0$ such that:
  $$
  f(\vw) \ge f(\vw^*) \quad \forall \vw \in \RR^d \text{ with } \|\vw - \vw^*\| \le \epsilon
  $$

- **Global Minimum**: $\hat{\vw}$ is a global minimizer if:
  $$
  f(\vw) \ge f(\hat{\vw}) \quad \forall \vw \in \RR^d
  $$
````

Local optimality describes the function’s behavior in a neighborhood, while global optimality describes the best solution across the entire space.



### Existence of Minimizers
Now let us talk about whether global or local optimal solution exists. Interestingly, even if the objective function is continuous, a global minimum may not be attained unless we impose additional structure. For example:

```{math}
\inf_{x \in \RR} \exp(-|x|) = 0
```
But the minimum is never achieved because $\exp(-|x|) \to 0$ as $|x| \to \infty$.

A standard result from real analysis helps clarify when minimizers exist:

````{admonition} Weierstrass Theorem
:class: important
 If $f: \RR^d \to \RR$ is continuous and its level set, defined below:

$$
\{ \vw \in \RR^d : f(\vw) \le f(\vw_0) \}
$$

is compact (i.e., closed and bounded) for some point $\vw_0$, then $f$ achieves its minimum on $\RR^d$.
````



### Optimality Conditions 
Now we know when the solutions exist, let's discuss how to identify them. Towards this end, we will discuss both *sufficient* and *necessary* conditions for a solution to be a local optimal solution. 





````{admonition} First-order and second-order necessary condition
:class: important
If $f$ is continuously differentiable, and let $\vw^*$ be a local minimizer. Then the following holds:

$$
\nabla f(\vw^*) = 0.
$$
This first-order necessary condition defines a (first-order) stationary points. They are candidates for global/local minmum, maximum solutions, as well as saddle points.

Further, suppose that $f$ is twice continuously differentiabe. Suppose $\vw^*$ is a local minimizer, then:

$$
\nabla f(\vw^*) = 0, \quad \nabla^2 f(\vw^*) \succeq 0.
$$

This second-order necessary condition, together with the first-order condition, defines a (second-order) stationary points. They are candidates for global and local minimum solutions.
````


```{admonition} Proof (First-order **Necessary** Condition)
:class: dropdown
:class: dropdown

First, consider the scalar case where $d=1$. 

By local minimality, there exists $\epsilon>0$ such that $f(x)\ge f(x^*)$ whenever $|x-x^*|\le \epsilon$.

- For $x\downarrow x^*$ (i.e., $x>x^*$ and $x\to x^*$), we have $f(x)-f(x^*)\ge 0$ and $x-x^*>0$, hence

$$
\lim_{x\downarrow x^*}\frac{f(x)-f(x^*)}{x-x^*}\;\ge\;0.
$$

- For $x\uparrow x^*$ (i.e., $x<x^*$ and $x\to x^*$), we have $f(x)-f(x^*)\ge 0$ and $x-x^*<0$, hence

$$
\lim_{x\uparrow x^*}\frac{f(x)-f(x^*)}{x-x^*}\;\le\;0.
$$

If $f$ is differentiable at $x^*$, the two one‑sided limits equal $f'(x^*)$. Therefore both inequalities imply that the following holds:

$$
f'(x^*)=0.
$$


Then let us consider the vector case where $d\ge 2$. Fix any direction $h\in\RR^d$ and define the scalar function $g(\alpha)=f(\,\vw^*+\alpha h\,)$.  Because $\vw^*$ is a local minimizer of $f$, we have that $\alpha=0$ is a local minimizer of $g$.  
Note that $f$ is $\mathcal C^1$, $g$ is differentiable and

$$
g'(0)=\nabla f(\vw^*)^\top h.
$$

Since $g(\cdot)$ is unconstrained and differentiable scalar function, then the fact that at $\alpha=0$ it achieves local minimum implies $g'(0)=0$ (see the first part of this proof), hence

$$
\nabla f(\vw^*)^\top d=0\quad\text{for all } d\in\RR^d.
$$

The only vector orthogonal to every direction $d$ is the zero vector, so $\nabla f(\vw^*)=\vzero$.

```


```{admonition} Proof (Second-Order **Necessary** Condition)
:class: dropdown


Again, first consider the scalar case where $d=1$. From the first-order necessary condition, $f'(x^*)=0$.  
Since $x^*$ is a local minimum solution, for $x$ sufficiently close to $x^*$ we have $f(x)\ge f(x^*)$. Consider

$$
q(x)\;=\;\frac{\,f(x)-f(x^*)-f'(x^*)(x-x^*)\,}{(x-x^*)^2}.
$$

Since $f'(x^*)=0$ and $f(x)\ge f(x^*)$, we get $q(x)\ge 0$ for all $x\neq x^*$ sufficiently close to $x^*$.  
Because $f$ is $\mathcal C^2$ at $x^*$, the limit exists and

$$
\lim_{x\to x^*} q(x)\;=\;\frac{1}{2}f''(x^*).
$$

Taking limits preserves the inequality, hence $\tfrac{1}{2}f''(x^*)\ge 0$, i.e., $f''(x^*)\ge 0$.


Then consider the vector case where $d\ge 2$. Fix any direction $h\in\RR^d$ and define $g(\alpha)=f(\vw^*+\alpha h)$.  
Since $\vw^*$ is a local minimizer of $f$, $\alpha=0$ is a local minimizer of the scalar $\mathcal C^2$ function $g$.  
From the first-order result, $g'(0)=\nabla f(\vw^*)^\top h=0$. By the chain rule,

$$
g''(0)= h^\top \nabla^2 f(\vw^*)\, h.
$$

From the first part of the proof, for scalar functions, if $\alpha^* =0$ is a local min of a twice differentiable scalar function, then $g''(0)\ge 0$. 
Because $h$ is arbitrary, $h^\top \nabla^2 f(\vw^*) h \ge 0$ for all $h$, i.e., $\nabla^2 f(\vw^*)\succeq 0$.
```


### Sufficient Optimality Conditions

````{admonition} Second-Order **Sufficient** Condition
:class: important
Let $f:\RR^d\to\RR$ be $\mathcal C^2$ near $\vw^*$.  
Supoose that $f(\cdot)$ is twice continuously differentiable, then if the following holds:

$$
\nabla f(\vw^*)=\vzero, \quad \nabla^2 f(\vw^*) \succ 0,
$$

then $\vw^*$ is a **strict local minimizer** of $f$.
````


````{admonition} Proof (Second-order Sufficient Conditions)
:class: dropdown
If $d=1$, by Taylor expansion, for $x$ close to $x^*$ there exists some $\xi$ between $x$ and $x^*$ such that

$$
f(x) = f(x^*) + f'(x^*)(x-x^*) + \tfrac12 f''(\xi)\,(x-x^*)^2.
$$
This is also known as the (scalar version of) **Mean Value Theorem (MVT)**.

From the first-order condition, $f'(x^*)=0$.  By continuity of $f''$ and the fact that $f''(x^*)>0$, there exists a neighborhood $U$ of $x^*$ such that $f''(\xi) \ge \mu > 0$ for all $\xi\in U$.  
Thus, for $x \in U\setminus\{x^*\}$,

$$
f(x) - f(x^*) \;\ge\; \tfrac{\mu}{2}(x-x^*)^2 \;>\; 0,
$$

which proves $x^*$ is a **strict local minimizer**.

If $d\ge 2$, by the multivariate Taylor’s theorem, for $\vh$ sufficiently small there exists $\theta\in(0,1)$ such that

$$
f(\vw^*+\vh) = f(\vw^*) + \nabla f(\vw^*)^\top \vh + \tfrac12 \vh^\top \nabla^2 f(\vw^*+\theta\vh)\, \vh.
$$

This is also known as the (vector version of) MVT.

From the first-order condition, $\nabla f(\vw^*)=\vzero$.   Since $\nabla^2 f(\vw^*) \succ 0$ and is continuous, there exists $\mu>0$ and a neighborhood $\mathcal N$ of $\vw^*$ such that $\nabla^2 f(\vw) \succeq \mu I$ for all $\vw\in\mathcal N$.  Therefore, for $\vh$ small with $\vw^*+\vh\in\mathcal N$,

$$
f(\vw^*+\vh) - f(\vw^*) \;\ge\; \tfrac{\mu}{2}\|\vh\|^2 \;>\; 0 \quad (\vh \neq 0),
$$

which proves $\vw^*$ is a strict local minimizer.
````

### Examples: Using Optimality Conditions

#### Example 1: Scalar quadratic (global minimum)
Let

$$
f(x) = (x-2)^2.
$$

- First-order condition:

$$
f'(x) = 2(x-2) = 0 \quad\Rightarrow\quad x^* = 2.
$$

- Second-order condition:

$$
f''(x) = 2 > 0.
$$

Conclusion: $x^*=2$ is a **strict local (and global) minimizer**.

---

#### Example 2: Scalar quartic (flat at minimum)
Let

$$
f(x) = x^4.
$$

- First-order condition: $f'(x) = 4x^3 = 0 \quad\Rightarrow\quad x^* = 0$.
- Second-order condition: $f''(x) = 12x^2 \quad\Rightarrow\quad f''(0) = 0$.
- Necessary conditions hold, but **sufficient** condition is inconclusive.
Checking directly: $f(x) \ge 0$ for all $x$, so $x^* = 0$ is indeed a **global minimum**.

---

#### Example 3: Scalar cubic (saddle / inflection point)
Let

$$
f(x) = x^3.
$$

- First-order: $f'(x) = 3x^2 = 0 \quad\Rightarrow\quad x^* = 0$.
- Second-order: $f''(x) = 6x \quad\Rightarrow\quad f''(0) = 0$. 
- Test inconclusive. Checking: $f(x) < 0$ for $x < 0$ and $f(x) > 0$ for $x > 0$. This implies that $x^* = 0$ is **not** a (local)minimum.

---

#### Example 4: Multivariate quadratic (positive definite Hessian)
Let

$$
f(\vw) = \tfrac12 \vw^\top Q \vw - \vb^\top \vw,
$$

where

$$
Q = \begin{bmatrix} 2 & 0 \\ 0 & 4 \end{bmatrix}, \quad \vb = \begin{bmatrix} 2 \\ 4 \end{bmatrix}.
$$

- First-order: $\nabla f(\vw) = Q\vw - \vb = 0 \quad\Rightarrow\quad \vw^* = Q^{-1} \vb = (1, 1)^\top$.
- Second-order: $\nabla^2 f(\vw) = Q \succ 0$ (positive definite).

Conclusion: $\vw^* = (1,1)^\top$ is a **strict local (and global) minimizer**.

---

#### Example 5: Multivariate saddle point
Let

$$
f(x_1, x_2) = x_1^2 - x_2^2.
$$

- First-order: $\nabla f = (2x_1, -2x_2)^\top = 0 \quad\Rightarrow\quad (x_1, x_2) = (0,0)$.
- Second-order: $\nabla^2 f = \begin{bmatrix} 2 & 0 \\ 0 & -2 \end{bmatrix}$. Noe that this matrix is neither a positive nor a negative definite matrix. 

Conclusion: $(0,0)$ is a **saddle point**, not a minimum.



### Why Are Optimality Conditions Important?
- They provide verification tools — necessary conditions rule out non-optimal points; sufficient conditions certify local optima.

- They are used to design stopping criteria in algorithms: e.g., when $\|\nabla f(\vw)\| \le \text{tolerance}$.

- They guide algorithmic design: most algorithms search for points satisfying $\nabla f(\vw) = 0$ (approximately).

- In convex problems, first-order conditions are sufficient, making them very powerful. More to come later in subsection **[xxxx]**


### When Does a Local Minimum Become a Global Minimum?

In general, a **local minimum** solution needs not be a **global minimum**.   However, for some special classes of optimization problems, *every* local minimum is guaranteed to be global.   Knowing these cases is important because it means **checking local optimality conditions is enough** to guarantee global optimality.

1) **Convex Problems**
2) **Unconstrained Quadratic Programs (UQPs)**  

---

````{admonition} Definition: Convex Problem
:class: tip
An **optimization problem** is called **convex** if:

1. The **feasible set** $\mathcal{W}$ is a **convex set**, meaning  
   for any $\vw_1, \vw_2 \in \mathcal{W}$ and any $\lambda \in [0,1]$,  

   $$
   \lambda \vw_1 + (1-\lambda)\vw_2 \in \mathcal{W}.
   $$

2. The **objective function** $f:\mathcal{W}\to\RR$ is **convex**, meaning  
   for all $\vw_1, \vw_2 \in \mathcal{W}$ and all $\lambda\in[0,1]$,  

   $$
   f\big(\lambda\vw_1 + (1-\lambda)\vw_2\big)
   \ \le\ \lambda f(\vw_1) + (1-\lambda) f(\vw_2).
   $$
````
---

**Special case (unconstrained):**  
If $\mathcal{W}=\RR^d$, a convex problem reduces to minimizing a convex function over the entire space.

---

- Intuitively, a convex function ‘holds water’.
- Further, if $f$ is second-order differentiable, then convexity ⇔ $\nabla^2 f(\vw) \succeq 0$ for all $\vw$ in the domain.  
- Linear functions are both convex and concave.



````{admonition} Convex Problems: Local ⇒ Global 
:class: tip
If $f:\RR^d\to\RR$ is **convex**, then **every local minimizer is a global minimizer**. 
 
Clearly, if $\bar{\vw}$ were local but not global, there exists $\vw$ with $f(\vw)<f(\bar{\vw})$. By convexity,

$$
f\big(c\,\bar{\vw}+(1-c)\vw\big)\le c\,f(\bar{\vw})+(1-c)f(\vw)< f(\bar{\vw}),
$$

contradicting local optimality at $\bar{\vw}$. 
````


````{admonition} UQPs: Local ⇒ Global (and uniqueness when $Q\succ 0$)
:class: tip
Consider the quadratic problem 

$$
f(\vw)=\tfrac12\,\vw^\top Q\,\vw+\vb^\top \vw + c,\quad Q=Q^\top.
$$

- **First-Order Necessary Condition:** $\nabla f(\vw)=Q\vw+\vb=0$. 
- **Second-Order Necessary Condition** $\nabla^2f(\vw) = Q\succeq 0$. 
- So every local optimal solution has to satisfy the above two conditions. 
- However, note that if $Q\succeq 0$, then the problem is convex, local min == global min. 
- The above necessary condition is also sufficient. 
- If $Q\succ 0$ (PD), the minimizer is **unique** (strong convexity). 
````

Convexity is a fundamental concept in optimization. Many powerful results — such as “local minima are global” — rely on it.

---

````{admonition} Definition: Convex Function
:class: tip
A function $f:\mathcal{W} \to \RR$ is **convex** if for all $\vw_1, \vw_2 \in \mathcal{W}$ and all $\lambda \in [0,1]$,
$$
f\big(\lambda\vw_1 + (1-\lambda)\vw_2\big) 
\ \le\ \lambda f(\vw_1) + (1-\lambda) f(\vw_2).
$$
If the inequality is strict whenever $\vw_1\neq \vw_2$ and $\lambda\in(0,1)$, $f$ is **strictly convex**.

Intuition: The line segment between any two points on the graph of $f$ lies **above** the graph.
