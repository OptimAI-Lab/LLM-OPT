# Gradient Methods 

Recall that our aim is to solve/tackle the following unconstrained optimization problem:
```{math}
\min_{\vw \in \RR^d} \quad f(\vw)
```
where $\vw$ is the optimization variable; $f: \RR^d \to \RR$ is the objective function. There are no explicit constraints on $\vw$. 

To make our life simpler, we assume that $f$ is (continuously) differentiable everywhere. As discussed in the last subsection, the task of tackling the optimization problem can be cast as searching for $\vw^*$ such that 
```{math}
\nabla f(\vw^*) = 0.
```

To this end, **gradient (a.k.a. first-order) methods** are efficient numerical methods for achieving the above goals. In its simplest form, the gradient method can be described as: 

````{prf:algorithm} Gradient Method
:label: alg:gradient

- **Input**: initial point $\vw^0$, stepsize sequence $\{ \gamma_k \}_{k \geq 0}$, max. no. of iterations $K$.

- For $k=0,1,2,..., K-1$,
```{math}
\vw^{k+1} = \vw^k - \gamma \nabla f( \vw^k).
```
- **Output**: last iterate $\vw^K$, or the solution sequence $\{ \vw^k \}_{k=1}^K$.
````

## Assumptions
